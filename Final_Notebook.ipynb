# Preproccesing Without Tokenization

### Preprocssing Functions

```python
import re
#removing those parts instead of replacing them with tags (<url> or <number>)
#because it is more fitting for summarization
```

```python
#remove urls because we wont need it for summarization
def remove_urls(text):
    return re.sub(r'https?:\/\/.*[\r\n\s]*', '', text, flags=re.MULTILINE)
```

```python
#normalize to lower case
def lower_case(text):
    return ''.join([w.lower() for w in text])
```

```python
##remove Email because we wont need it for summarization
def remove_email(text):
    return re.sub(r'\S*@\S*\s?', '', text, flags=re.MULTILINE)
```

```python
# remove tags because we wont need it for summarization
def remove_tags(text):
    # Remove tags like <file_photo>, <file_gif>, etc.
    text = re.sub(r'<[^>]+>', '', text)
    return text
```

```python
##remove english because we want the text to be arabic
def remove_english_characters(text):
    # Remove all English alphabetic characters (both uppercase and lowercase)
    text = re.sub(r'[A-Za-z]', '', text)
    return text
```

```python
#stem both the input and the output

from nltk.stem import SnowballStemmer

s_stemmer = SnowballStemmer('arabic')
```

```python
#remove the new line and tags that are not needed
def normalize(text):
    text = text.replace('\n', '')
    text = text.replace('\r', ' ')
    text = text.replace('\t', ' ')
    text = text.replace('+', ' ')
    return text
```

```python
def normalize_arabic(text):
    # Remove diacritics
    diacritics = re.compile(r'[\u064B-\u065F]')
    text = re.sub(diacritics, '', text)
    # Normalize letters
    text = re.sub(r'[إأآا]', 'ا', text)
    text = re.sub(r'ى', 'ي', text)
    text = re.sub(r'ؤ', 'و', text)
    text = re.sub(r'ئ', 'ي', text)
    # Remove tatweel
    text = re.sub(r'ـ', '', text)
    return text
```

```python
def clean_text(text):
    text = re.sub(r'\s+', ' ', text)    # Remove extra spaces
    return text
```

```python
def preprocess_text(text):
    text = remove_urls(text)
    text = remove_email(text)
    text = remove_tags(text)
    text = remove_english_characters(text)
    text = normalize(text)
    text = lower_case(text)
    normalize_arabic(text)
    clean_text(text)
    # text = stem_text(text)
    return text
    #won't stem now will stem after tokenization
```

```python
from transformers import T5Tokenizer, AutoModelForSeq2SeqLM

tokenizer = T5Tokenizer.from_pretrained("UBC-NLP/AraT5v2-base-1024")
model = AutoModelForSeq2SeqLM.from_pretrained("UBC-NLP/AraT5v2-base-1024")
```

```python
ar_prompt="عاصمة ألمانيا هي <extra_id_0> "
input_ids = tokenizer(ar_prompt, return_tensors="pt").input_ids
outputs = model.generate(input_ids)
print("Tokenized input:", tokenizer.tokenize(ar_prompt))
print("Decoded output:", tokenizer.decode(outputs[0], skip_special_tokens=True))
```

### Input normal and output normal

```python
max_input_length = 384
max_target_length = 96

def tokenize_fn(batch):
    inputs = ["لخص الحوار: " + d for d in batch["dialogue"]]

    model_inputs = tokenizer(
        inputs,
        truncation=True,
        max_length=max_input_length,
    )

    labels = tokenizer(
        batch["summary"],
        truncation=True,
        max_length=max_target_length,
    )

    model_inputs["labels"] = labels["input_ids"]
    return model_inputs
```

```python
from datasets import DatasetDict
tokenized_splits = {}

for split in ds:
    tokenized_splits[split] = ds[split].map(
        tokenize_fn,
        batched=True,
        batch_size=16,      # small batch to prevent OOM
        remove_columns=ds[split].column_names,
        keep_in_memory=False
    )

tokenized_ds = DatasetDict(tokenized_splits)
```

```python
import evaluate

rouge = evaluate.load("rouge")
```

```python
def compute_metrics(eval_preds):
    preds, labels = eval_preds

    # Decode predictions
    decoded_preds = tokenizer.batch_decode(
        preds, skip_special_tokens=True
    )

    # Replace -100 in labels (Trainer uses this internally)
    labels = [
        [(token if token != -100 else tokenizer.pad_token_id) for token in label]
        for label in labels
    ]

    decoded_labels = tokenizer.batch_decode(
        labels, skip_special_tokens=True
    )

    result = rouge.compute(
        predictions=decoded_preds,
        references=decoded_labels,
    )

    # Return ROUGE-1, ROUGE-2, ROUGE-L
    return {
        "rouge1": result["rouge1"],
        "rouge2": result["rouge2"],
        "rougeL": result["rougeL"],
    }
```

```python
from transformers import DataCollatorForSeq2Seq

data_collator = DataCollatorForSeq2Seq(
    tokenizer=tokenizer,
    model=model,
)
```

```python
from transformers import TrainingArguments

training_args = TrainingArguments(
    output_dir="/content/arat5_out",
    per_device_train_batch_size=1,
    gradient_accumulation_steps=8,
    fp16=True,
    learning_rate=3e-5,
    num_train_epochs=2,
    save_total_limit=1,
    logging_steps=100,
)
```

```python
from transformers import Trainer

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_ds["train"].select(range(600)),
    tokenizer=tokenizer,
    data_collator=data_collator,
)
```

```python
trainer.train()
```

```python
import evaluate
rouge = evaluate.load("rouge")
```

```python
import torch

val_subset = tokenized_ds["validation"].select(range(40))  # 🔹 small subset
preds = []
refs = []

model.eval()
for i, batch in enumerate(val_subset):
    input_ids = torch.tensor(batch["input_ids"]).unsqueeze(0).to(model.device)
    attention_mask = torch.tensor(batch["attention_mask"]).unsqueeze(0).to(model.device)

    with torch.no_grad():
        output = model.generate(
            input_ids=input_ids,
            attention_mask=attention_mask,
            max_length=96,
            num_beams=1,
        )

    preds.append(tokenizer.decode(output[0], skip_special_tokens=True))

    labels = [x if x != -100 else tokenizer.pad_token_id for x in batch["label"]]
    refs.append(tokenizer.decode(labels, skip_special_tokens=True))

    torch.cuda.empty_cache()  # 🔹 clear GPU memory each loop
```

```python
results = rouge.compute(predictions=preds, references=refs)
print(results)
```

```python
import evaluate

rouge = evaluate.load("rouge")


def compute_metrics(eval_preds):
    preds, labels = eval_preds

    # Decode predictions
    decoded_preds = tokenizer.batch_decode(
        preds, skip_special_tokens=True
    )

    # Replace -100 in labels (Trainer uses this internally)
    labels = [
        [(token if token != -100 else tokenizer.pad_token_id) for token in label]
        for label in labels
    ]

    decoded_labels = tokenizer.batch_decode(
        labels, skip_special_tokens=True
    )

    result = rouge.compute(
        predictions=decoded_preds,
        references=decoded_labels,
    )

    # Return ROUGE-1, ROUGE-2, ROUGE-L
    return {
        "rouge1": result["rouge1"],
        "rouge2": result["rouge2"],
        "rougeL": result["rougeL"],
    }
```
